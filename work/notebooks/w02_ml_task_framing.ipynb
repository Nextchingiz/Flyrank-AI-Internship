## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# This is a ranking problem, not classification or clustering. The real decision is "which declining content should an editor fix first, given they don't have time for everything." That's a which-ones-first question, so the output needs to be an ordered list, not a single yes/no. The model scores each content item by how urgently it needs a refresh, and editors work down that list from the top, starting with whatever's flagged most urgent.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# I won't use is_declining_label as it is, since it's derived from trend_direction, which comes from trend_pct. That's someone else's rule, not something observed. Using it directly means the model just learns to reproduce that rule instead of learning real decline patterns.
# Instead, the target should be built from a real future outcome. Whether a content piece's performance (traffic, engagement, position) actually drops over a defined future window, measured directly from the data rather than from the pre built label. That keeps the target observed, not defined.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# Editors only have time to work through a small number of items, not the whole list, so precision at the top of the ranking is what matters most. Precision@K measures, out of the top K flagged items an editor would actually look at, how many were truly declining. This lines up with what matters most for this business: avoiding false alarms. A low precision@K means editors keep getting sent to fix things that were not actually broken, which is the exact cost I decided was worse earlier. Recall matters less here, since missing a few real declines is a smaller cost than wasting editor time again and again.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# One row is one piece of content, specifically one pseudonymized content item belonging to one of FlyRank's 32 clients, described by its trailing 90 day performance metrics. This matches the decision from the frame, since the editor is not deciding at the client level or the query level, they are deciding which individual content items to refresh first. So the model needs to score and rank at the content item level, one row per item, not aggregated up to the client or down to individual search queries.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# A single if-statement, like flagging anything with trend_pct below some threshold, would not work across FlyRank's 32 clients. Different clients have different normal ranges and different baselines, so a drop that is routine for one client could be alarming for another. Content type matters too, since signals show up differently depending on the kind of content, and missingness itself varies by content type. One fixed threshold cannot flex for any of that. A model can learn client specific and content type specific patterns from the data itself, instead of forcing the same rule onto every client.
```
